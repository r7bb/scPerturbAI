# Baseline Models

Three baselines, evaluated with strict **perturbation-level** train/test splits (see `src/data/splits.py`): a perturbation's cells never appear in both train and test, and two-gene combinations are further stratified by how many of their two genes were ever seen as single-gene perturbations during training (0/1/2 unseen).

- **No-perturbation**: always predict zero change.
- **Mean-shift**: always predict the average delta across training perturbations.
- **Ridge regression**: gene-pair embedding → expression delta, compared with two embedding variants (see `05_deep_learning.ipynb` for what the embeddings are).

```python
import pandas as pd
pd.set_option('display.width', 200)
metrics = pd.read_csv('../results/tables/baseline_metrics.csv')
metrics.round(3)
```

**Reading this table:** `pearson_delta` is the correlation between predicted and true expression change across all 2000 modeled genes; `de_pearson_delta` and `direction_accuracy` restrict to the top-50 genes by true |delta| (the genes that actually matter). `test_combo_2_unseen` has only 5 perturbations — treat it as a qualitative signal, not a precise estimate.

Even Ridge regression with a good gene-pair representation clears the mean-shift baseline by a wide margin on `test_combo_0_unseen` (0.65–0.90 vs. 0.57), confirming the embeddings carry real signal rather than the model just learning a global average response.
